```python
import pandas as pd

timestamps = [
  "2024-05-19 14:30:56.591030836+00:00",
  "2024-05-20 02:36:39.816505671+00:00",
  "2024-05-21 14:55:16.659343243+00:00",
  "2024-05-22 03:01:23.930391312+00:00"
]

timestamp_series = pd.Series(timestamps)
timestamp_series_utc = pd.to_datetime(timestamp_series).dt.tz_convert('UTC')

print(timestamp_series_utc)

```

```python
# find timedelta
print(timestamp_series_utc.diff().dt.total_seconds())

print(timestamp_series_utc.diff())
```

```python
def format_timedelta(td):
    days = td.days
    seconds = td.seconds
    return f"{days}d,{seconds}s"
```

```python
timedelta_series = timestamp_series_utc.diff()

formatted_series = timedelta_series.apply(lambda x: format_timedelta(pd.Timedelta(x)))

print(formatted_series)
```
