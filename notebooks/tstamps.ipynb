```python
import pandas as pd
from datetime import timedelta

def tstamp2time(t):
  time, frac = t.split('.')
  time = pd.to_datetime(int(time), unit='s').tz_localize('UTC')
  frac = int(f"{frac:<06}")
  time = time + timedelta(microseconds=frac)
  return time

```

```python
tstamps = [1719510239.483,1719513240.628,1719515340.56,1719522240.518,1719526140.401,1719528240.416,1719530340.345]

for t in tstamps:
  tstamp2time(str(t))
```
